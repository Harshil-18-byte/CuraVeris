# 02. Preprocessing & Normalization Pipeline
Cleans OCR noise, standardizes hospital clinical item names, extracts numeric rates, quantities, and GST.

```python
import json
import pandas as pd

# Extract line items into flat dataframe
items = []
with open('../data/processed/synthetic_bills.jsonl', 'r') as f:
    for line in f:
        bill = json.loads(line)
        for it in bill.get('items', []):
            it['hospital_name'] = bill['hospital_name']
            it['diagnosis'] = bill['diagnosis']
            items.append(it)

df_items = pd.DataFrame(items)
print(f"Total line items: {len(df_items)}")
df_items.head()
```
